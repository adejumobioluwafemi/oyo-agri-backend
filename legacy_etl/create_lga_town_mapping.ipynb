```python
import csv
import re

# Raw text data
raw_data = """
"LGA" "District / area" "Postal code" "Towns" Afijio Akinmorin / Jobele 211102 Aba Ibadan; Aba Kasia; Akinmorin; Alapadi; Anfelerin; Bali; Basi; Ekewa; Eleko; Elepe; Elesu; Elewebe; Fadare; Idiito; Ijefon; Ile Olufi; Imeleke; Iseke; Jagun; Jobele; Jokan; Kolobo; Lanipekun; Lapiti; Oke; Olokiti; Oniyanrin; Sulu Afijio Awe 211106 Aba Bara; Aba Ogunremi; Aba Olori; Aiyekale; Ajegunle; Apaara; Asipa; Awe; Emiabata; Idi Agbon; Idi Orpa; Idimangoro; Ijado; Isale Apata; Jabata; Kiyesemi; Kosobo; Laraniyi; Moromipin; Olukolu; Oluode; Onfunfun
...
"""

def parse_oyo_data(raw_text):
    """
    Parse the raw Oyo LGA data into a structured CSV format.
    
    Args:
        raw_text: String containing the raw data
    
    Returns:
        List of dictionaries with keys: lga, district_area, postalcode, town
    """
    results = []
    
    # Remove the header line
    lines = raw_text.strip().split('\n')
    
    for line in lines:
        if not line.strip() or line.startswith('"LGA"'):
            continue
        
        # Pattern to match: LGA District/Area PostalCode Towns (semicolon-separated)
        # Extract LGA name (word at start)
        parts = line.strip().split()
        
        if len(parts) < 3:
            continue
        
        # Find LGA (first word)
        lga = parts[0]
        
        # Find postal code (6-digit number)
        postal_code = None
        postal_idx = -1
        for i, part in enumerate(parts):
            if re.match(r'^\d{6}$', part):
                postal_code = part
                postal_idx = i
                break
        
        if not postal_code:
            continue
        
        # District/area is between LGA and postal code
        district_area = ' '.join(parts[1:postal_idx])
        
        # Towns are after postal code, separated by semicolons
        towns_text = ' '.join(parts[postal_idx + 1:])
        towns = [town.strip() for town in towns_text.split(';')]
        
        # Create a row for each town
        for town in towns:
            if town:  # Skip empty strings
                results.append({
                    'lga': lga,
                    'district_area': district_area,
                    'postalcode': postal_code,
                    'town': town
                })

        #if towns:  # Skip empty strings
        #    results.append({
        #        'lga': lga,
        #        'district_area': district_area,
        #        'postalcode': postal_code,
        #        'town': towns
        #    })
    return results

def save_to_csv(data, filename='oyo_lga_town_map.csv'):
    """
    Save the parsed data to a CSV file.
    
    Args:
        data: List of dictionaries
        filename: Output CSV filename
    """
    if not data:
        print("No data to save")
        return
    
    with open(filename, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=['lga', 'district_area', 'postalcode', 'town'])
        writer.writeheader()
        writer.writerows(data)
    
    print(f"Data saved to {filename}")
    print(f"Total rows: {len(data)}")
```

```python
# Read from file or use the raw_data string
with open('raw_town_lga_data.txt', 'r', encoding='utf-8') as f:
    raw_text = f.read()

parsed_data = parse_oyo_data(raw_text)

save_to_csv(parsed_data)

print("\nFirst 10 rows:")
for i, row in enumerate(parsed_data[:10], 1):
    print(f"{i}. {row}")
```
